## Train and evaluate

The holdout retains the historical stratified split. Five-fold validation uses training rows only. The imputer and scaler fit within each fold and on the final training partition.

```python
model, report = fit_and_evaluate()
print(json.dumps(report, indent=2))
```

## Score the supplied example

The record uses named features. Class 1 denotes the positive benchmark label. The decision score is the signed SVC margin.

```python
record = json.loads(Path("examples/record.json").read_text(encoding="utf-8"))
print(predict(model, record))
```
